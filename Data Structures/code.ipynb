### Array

```python
# Creation
import array as ar

arr = ar.array('i', [1,2,3,4,5])
print(arr) 
```

```python
# Access
arr[4]
```

```python
#Insert
arr.insert(0,6)
print(arr)
```

```python
#Delete
arr.pop(0)
print(arr)
```

```python
#Search
a=2
for i in range(0,len(arr)):
    if arr[i] == a:
        print(i)
        break
```

## Stack

```python
# Stack can be implemented in python using List and deque
stack = [1,2,3,4] # LIFO

print(stack[-1]) # Accessing
stack.append(5) # Insertion
print(stack)
print(stack[-1]) # Access

stack.pop() # Delete
print(stack)
```
